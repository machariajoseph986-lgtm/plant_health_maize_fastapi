## 1.1 Problem Statement

Maize is an important crop whose production can be affected by plant diseases. Early and accurate identification of disease symptoms is important because incorrect or delayed identification can affect the effectiveness of subsequent management decisions.

However, identifying maize diseases from leaf images can be challenging because different diseases may produce visually similar symptoms, while variations in image quality, lighting, leaf condition, and disease severity can further complicate classification.

This project therefore investigates whether image-based machine learning can reliably classify maize leaf images into the defined disease and healthy categories.

The project further investigates whether the resulting classification can support the provision of useful and trustworthy information to a user based on the predicted condition.

The classification task focuses on four categories:

- **Blight**
- **Common_Rust**
- **Gray_Leaf_Spot**
- **Healthy**

The system is intended to provide **decision support and information**, rather than present a machine-learning prediction as a confirmed diagnosis. Model performance, confidence, uncertainty, and the reliability of the information associated with the prediction are therefore important considerations throughout the project.

````python
# Purpose: Verify that the training, validation, and test directories contain the expected four maize classification classes.

EXPECTED_CLASSES = [
    "Blight",
    "Common_Rust",
    "Gray_Leaf_Spot",
    "Healthy"
]

split_directories = {
    "Training": TRAIN_DIR,
    "Validation": VALIDATION_DIR,
    "Test": TEST_DIR
}

print("DATASET SPLIT STRUCTURE VERIFICATION")
print("=" * 80)

all_valid = True

for split_name, split_dir in split_directories.items():
    print(f"\n{split_name.upper()}")
    print("-" * 80)

    if not split_dir.exists():
        print(f"ERROR: Directory not found: {split_dir}")
        all_valid = False
        continue

    found_classes = sorted(
        [item.name for item in split_dir.iterdir() if item.is_dir()]
    )

    print(f"Found classes: {found_classes}")

    for class_name in EXPECTED_CLASSES:
        class_dir = split_dir / class_name

        if class_dir.exists() and class_dir.is_dir():
            print(f"  ✓ {class_name}")
        else:
            print(f"  ✗ {class_name} — MISSING")
            all_valid = False

    unexpected_classes = [
        class_name
        for class_name in found_classes
        if class_name not in EXPECTED_CLASSES
    ]

    if unexpected_classes:
        print(f"  WARNING: Unexpected directories: {unexpected_classes}")
        all_valid = False

print("\n" + "=" * 80)

if all_valid:
    print("STATUS: Dataset split structure is valid.")
else:
    print("STATUS: Dataset split structure requires investigation.")
````

````python
# Purpose: Count the images in each class for the training, validation, and test datasets.

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

print("DATASET SPLIT IMAGE COUNTS")
print("=" * 90)

split_counts = {}
grand_total = 0

for split_name, split_dir in split_directories.items():
    print(f"\n{split_name.upper()}")
    print("-" * 90)

    split_counts[split_name] = {}
    split_total = 0

    for class_name in EXPECTED_CLASSES:
        class_dir = split_dir / class_name

        image_files = [
            file_path
            for file_path in class_dir.rglob("*")
            if file_path.is_file()
            and file_path.suffix.lower() in IMAGE_EXTENSIONS
        ]

        count = len(image_files)
        split_counts[split_name][class_name] = count
        split_total += count

        print(f"{class_name:<20} {count:>6} images")

    print("-" * 90)
    print(f"{'TOTAL':<20} {split_total:>6} images")

    grand_total += split_total

print("\n" + "=" * 90)
print(f"COMBINED DATASET TOTAL: {grand_total} images")
````

````python
# Purpose: Calculate and display the class distribution percentages for the training, validation, and test datasets.

print("DATASET CLASS DISTRIBUTION")
print("=" * 100)

for split_name, class_counts in split_counts.items():
    split_total = sum(class_counts.values())

    print(f"\n{split_name.upper()}")
    print("-" * 100)
    print(f"{'Class':<20} {'Images':>10} {'Percentage':>15}")

    for class_name in EXPECTED_CLASSES:
        count = class_counts[class_name]
        percentage = (count / split_total) * 100

        print(f"{class_name:<20} {count:>10} {percentage:>14.2f}%")

    print("-" * 100)
    print(f"{'TOTAL':<20} {split_total:>10} {'100.00%':>15}")
````

````python
# Purpose: Verify image integrity, dimensions, and color modes across the complete dataset split.

from PIL import Image
from collections import Counter

print("IMAGE INTEGRITY AND FORMAT VERIFICATION")
print("=" * 100)

all_image_paths = []

for split_name, split_dir in split_directories.items():
    for class_name in EXPECTED_CLASSES:
        class_dir = split_dir / class_name

        for file_path in class_dir.rglob("*"):
            if file_path.is_file() and file_path.suffix.lower() in IMAGE_EXTENSIONS:
                all_image_paths.append(file_path)

print(f"Total image files found: {len(all_image_paths)}")

corrupted_images = []
image_dimensions = Counter()
image_modes = Counter()

for image_path in all_image_paths:
    try:
        with Image.open(image_path) as img:
            img.verify()

        with Image.open(image_path) as img:
            image_dimensions[img.size] += 1
            image_modes[img.mode] += 1

    except Exception as error:
        corrupted_images.append((image_path, str(error)))

print("\nIMAGE INTEGRITY")
print("-" * 100)

if corrupted_images:
    print(f"Corrupted/unreadable images: {len(corrupted_images)}")
    for image_path, error in corrupted_images[:10]:
        print(f"  {image_path} -> {error}")
else:
    print("Corrupted/unreadable images: 0")
    print("STATUS: All image files passed the integrity check.")

print("\nIMAGE DIMENSIONS")
print("-" * 100)
print(f"Different image dimensions found: {len(image_dimensions)}")

print("\nMost common dimensions:")
for dimension, count in image_dimensions.most_common(10):
    percentage = (count / len(all_image_paths)) * 100
    print(f"  {dimension}: {count} images ({percentage:.2f}%)")

print("\nIMAGE COLOR MODES")
print("-" * 100)

for mode, count in image_modes.most_common():
    percentage = (count / len(all_image_paths)) * 100
    print(f"  {mode}: {count} images ({percentage:.2f}%)")

print("\n" + "=" * 100)
````

# 3. Data Preparation

The Data Preparation phase transforms the verified dataset into a consistent form suitable for model development.

Based on the findings from the Data Understanding phase, the dataset is already structurally valid and contains no corrupted images. However, the images have different original dimensions and a small number use color modes other than RGB. Therefore, standardized preprocessing is required before model training.

The preparation process for this project consists of:

1. Standardizing images to a consistent input size.
2. Converting images to RGB where necessary.
3. Normalizing pixel values according to the requirements of each model.
4. Applying data augmentation to the training data only.
5. Preserving the validation and test datasets without augmentation.
6. Using the existing stratified train, validation, and test split.
7. Applying class weights during model training to account for the under-represented Gray_Leaf_Spot class.

The same authoritative dataset split will be maintained throughout the model experiments so that Custom Baseline, Tuned Custom Baseline, and MobileNetV2 Transfer Learning can be compared fairly.

````python
# PURPOSE: Define the standardized image-preparation configurations used by the Custom Baseline, Tuned Custom Baseline, and MobileNetV2 Transfer Learning experiments.

# Standard image sizes used in the model experiments.
IMG_SIZE_V1 = (256, 256)
IMG_SIZE_V2 = (320, 320)
IMG_SIZE_V3 = (320, 320)

# Batch size used for dataset pipelines.
BATCH_SIZE = 32

# Reproducibility seed.
SEED = 42

# Custom Baseline / Tuned Custom Baseline (custom CNN) preprocessing.
# Pixel values are scaled from [0, 255] to [0, 1].
V1_V2_NORMALIZATION = "rescale_1_over_255"

# MobileNetV2 preprocessing.
# MobileNetV2 preprocess_input scales pixel values to approximately [-1, 1].
V3_NORMALIZATION = "MobileNetV2 preprocess_input"

print("DATA PREPARATION CONFIGURATION")
print("=" * 80)

print(f"Custom Baseline image size:          {IMG_SIZE_V1}")
print(f"Tuned Custom Baseline image size:          {IMG_SIZE_V2}")
print(f"MobileNetV2 Transfer Learning image size:          {IMG_SIZE_V3}")
print(f"Batch size:             {BATCH_SIZE}")
print(f"Random seed:            {SEED}")

print("\nNORMALIZATION")
print("-" * 40)
print(f"Custom Baseline/Tuned Custom Baseline: {V1_V2_NORMALIZATION}")
print(f"MobileNetV2 Transfer Learning:    {V3_NORMALIZATION}")

print("\nSTATUS: Data preparation configuration defined.")
````

````python
# PURPOSE: Demonstrate standardized RGB conversion and resizing for model-ready image input.

from PIL import Image

# Define the authoritative project classes.
CLASS_NAMES = [
    "Blight",
    "Common_Rust",
    "Gray_Leaf_Spot",
    "Healthy"
]

# Select one representative image from the training dataset.
sample_image_path = next(
    path
    for class_name in CLASS_NAMES
    for path in (TRAIN_DIR / class_name).iterdir()
    if path.is_file() and path.suffix.lower() in {".jpg", ".jpeg", ".png"}
)

# Open the original image.
original_image = Image.open(sample_image_path)

# Standardize color mode to RGB.
rgb_image = original_image.convert("RGB")

# Resize to the Custom Baseline model input size.
standardized_image = rgb_image.resize(IMG_SIZE_V1)

print("IMAGE STANDARDIZATION VERIFICATION")
print("=" * 80)

print(f"Sample image:       {sample_image_path.name}")
print(f"Original size:      {original_image.size}")
print(f"Original mode:      {original_image.mode}")
print(f"Standardized size:  {standardized_image.size}")
print(f"Standardized mode:  {standardized_image.mode}")

assert standardized_image.size == IMG_SIZE_V1
assert standardized_image.mode == "RGB"

print("\nSTATUS: Image successfully standardized to RGB and 256 × 256.")
````

### Image Preprocessing Methodology

Image preprocessing is required to convert the heterogeneous source images into a consistent representation suitable for deep learning.

First, all images are converted to RGB to ensure that the model receives three consistent color channels regardless of the original image color mode.

Second, images are resized to the input dimensions required by each model. Custom Baseline uses 256 × 256 pixels, while Tuned Custom Baseline and MobileNetV2 Transfer Learning use 320 × 320 pixels.

Third, pixel-value normalization is applied according to the model architecture. The custom CNN experiments (Custom Baseline and Tuned Custom Baseline) scale pixel values from the original [0, 255] range to [0, 1]. The MobileNetV2 experiment (MobileNetV2 Transfer Learning) uses the preprocessing function provided for MobileNetV2, producing values in approximately the [-1, 1] range.

Data augmentation is applied only to the training data. This increases variation in the training examples while keeping the validation and test sets unchanged for objective evaluation.

The validation and test datasets therefore represent unseen data and are not augmented. This separation helps ensure that reported model performance reflects generalization rather than exposure to artificially modified evaluation images.

````python
# PURPOSE: Create a reusable function for RGB conversion, resizing, and array conversion.
import numpy as np

def prepare_image(image_path, target_size=IMG_SIZE_V1):
    """
    Load an image and standardize it for model input.

    Steps:
    1. Load the image.
    2. Convert it to RGB.
    3. Resize it to the requested target size.
    4. Convert it to a NumPy array.
    """

    image = Image.open(image_path).convert("RGB")
    image = image.resize(target_size)

    image_array = np.asarray(image, dtype=np.float32)

    return image_array


# Verify the reusable function using the representative image.
prepared_sample = prepare_image(
    sample_image_path,
    target_size=IMG_SIZE_V1
)

print("REUSABLE IMAGE PREPROCESSING FUNCTION")
print("=" * 80)

print(f"Output shape: {prepared_sample.shape}")
print(f"Output dtype: {prepared_sample.dtype}")
print(f"Pixel minimum: {prepared_sample.min():.1f}")
print(f"Pixel maximum: {prepared_sample.max():.1f}")

assert prepared_sample.shape == (256, 256, 3)
assert prepared_sample.dtype == np.float32
assert 0 <= prepared_sample.min() <= 255
assert 0 <= prepared_sample.max() <= 255

print("\nSTATUS: Reusable image preparation function verified.")
````

### Data Augmentation: Purpose and Application

Data augmentation was used to increase the variety of training examples available to the CNN without creating new independent observations. The purpose was to help the model learn disease patterns that remain useful when the same disease appears under different image orientations, scales, or lighting conditions.

The augmentation transformations used during model training were:

| Augmentation        | Purpose                                                              |
| ------------------- | -------------------------------------------------------------------- |
| Horizontal flip     | Provides variation in leaf orientation                               |
| Rotation            | Helps the model handle different camera and leaf orientations        |
| Zoom                | Helps the model handle differences in the apparent scale of the leaf |
| Contrast adjustment | Helps the model handle differences in lighting conditions            |

The augmentation was applied **only to the training data**. The validation and test sets were kept unaugmented so that they remained independent and could provide a fair evaluation of how well the model generalizes to unseen images.

### How the Model Sees an Augmented Image

The CNN does not need to know that an image has been augmented. During training, it receives the transformed image as an input together with the original class label.

For example, if an original image belongs to the `Blight` class, a horizontally flipped, slightly rotated, zoomed, or contrast-adjusted version of that image is still presented to the model with the label `Blight`.

Therefore, augmentation provides the model with different visual variations of the same underlying training example while preserving the disease class.

### Data Leakage Prevention

To prevent data leakage, the dataset was split into training, validation, and test sets **before augmentation was applied**.

The workflow was:

**Original dataset → Train / Validation / Test split → Augment training data only → Model training**

This order is important. If an original image and one of its augmented versions were placed in different datasets, the model could effectively see information from the same original image during training and evaluation. This would make the evaluation less reliable.

By splitting first and augmenting only the training set, the validation and test sets remain independent from the augmented training examples.

The augmented images therefore improve training diversity and are not counted as additional independent images in the original dataset. The cleaned dataset remains **4,184 real images**.

## 4.1 Custom Baseline (Custom CNN)

Custom Baseline is the project's baseline convolutional neural network developed for four-class maize disease and health classification.

The model uses standardized 256 × 256 RGB images and performs four-class classification across:

- Blight
- Common_Rust
- Gray_Leaf_Spot
- Healthy

The Custom Baseline pipeline uses pixel normalization by dividing pixel values by 255, training-only data augmentation, and class weights to account for the observed class imbalance.

The custom CNN architecture was selected as a balanced architecture rather than an extremely shallow or unnecessarily deep network. It uses convolutional feature extraction followed by pooling and regularization before the final classification layer.

The final output layer uses four neurons with softmax activation, producing a probability distribution across the four maize classes.

Custom Baseline serves as the reference baseline for subsequent experiments. Tuned Custom Baseline and MobileNetV2 Transfer Learning are therefore evaluated against the same authoritative test set so that any performance improvement or degradation can be interpreted relative to the established baseline.

The trained Custom Baseline model and its associated training artifacts are preserved in the project repository and are not overwritten by the subsequent experiments.

````python
# PURPOSE: Reconstruct the exact Custom Baseline CNN architecture from the project's original cnn_model.py definition.

import tensorflow as tf

IMAGE_SIZE_V1 = 256
NUM_CHANNELS_V1 = 3
NUM_CLASSES_V1 = 4

model_v1_rebuilt = tf.keras.Sequential([

    # Input
    tf.keras.layers.Input(
        shape=(IMAGE_SIZE_V1, IMAGE_SIZE_V1, NUM_CHANNELS_V1)
    ),

    # Convolutional Block 1
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Convolutional Block 2
    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Convolutional Block 3
    tf.keras.layers.Conv2D(
        filters=128,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Convolutional Block 4
    tf.keras.layers.Conv2D(
        filters=256,
        kernel_size=(3, 3),
        activation="relu",
        padding="same"
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Feature aggregation
    tf.keras.layers.GlobalAveragePooling2D(),

    # Classification head
    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),
    tf.keras.layers.Dropout(0.4),

    # Output
    tf.keras.layers.Dense(
        NUM_CLASSES_V1,
        activation="softmax"
    )

], name="maize_disease_cnn")


print("Custom Baseline REBUILT ARCHITECTURE")
print("=" * 80)

model_v1_rebuilt.summary()

print("\nSTATUS: Custom Baseline architecture reconstructed successfully.")
````

## MobileNetV2 Transfer Learning Experiment

The MobileNetV2 Transfer Learning experiment introduced a pretrained MobileNetV2 architecture to determine whether transfer learning could improve maize disease classification compared with the custom CNN models evaluated in Custom Baseline and Tuned Custom Baseline.

MobileNetV2 was initialized with ImageNet-pretrained weights and adapted to the four maize disease classes: Blight, Common_Rust, Gray_Leaf_Spot, and Healthy.

The experiment used 320×320 RGB images and MobileNetV2-specific preprocessing. Moderate data augmentation was applied only to the training data using horizontal flipping, small rotations, zooming, and contrast adjustment. Validation and test images were not augmented.

MobileNetV2 Transfer Learning was evaluated in two stages:

1. **Stage 1 — Frozen transfer learning:** The pretrained MobileNetV2 feature extractor was kept frozen while a new classification head was trained.
2. **Stage 2 — Fine-tuning:** Selected later MobileNetV2 layers were unfrozen and fine-tuned using a substantially lower learning rate while Batch Normalization layers remained frozen.

Both stages used the same stratified train, validation, and test split established during data preparation. This ensured that the MobileNetV2 Transfer Learning results could be compared fairly with Custom Baseline and Tuned Custom Baseline.

The purpose of MobileNetV2 Transfer Learning was not simply to obtain a higher accuracy, but to determine whether transfer learning and controlled fine-tuning could improve generalization and reduce the classification difficulty observed between Blight and Gray_Leaf_Spot.

## Confidence and Uncertainty Evaluation

Overall classification accuracy does not indicate how confident the model is in individual predictions. For this project, prediction confidence is therefore considered alongside classification performance.

The confidence analysis evaluates whether predictions can be separated into higher-confidence accepted predictions and lower-confidence uncertain predictions.

The project's confidence policy uses a **70% acceptance threshold**. Predictions with confidence below this threshold are treated as uncertain rather than being presented as confident disease identifications.

This approach is particularly important for the Blight and Gray_Leaf_Spot classes because these diseases showed the greatest remaining visual overlap during error analysis.

The confidence mechanism is intended as a safety layer rather than a guarantee of correctness. A high-confidence prediction can still be wrong, while an uncertain prediction may still correspond to the correct class.

Therefore, the system communicates predictions as indications rather than confirmed diagnoses and uses uncertainty handling to reduce the risk of presenting low-confidence predictions as reliable diagnoses.

````python
# PURPOSE: Record the verified Custom Baseline confidence-threshold results used by the project's uncertainty policy.

confidence_threshold_results = pd.DataFrame({
    "Threshold": ["50%", "60%", "70%", "75%", "80%", "85%", "90%", "95%"],
    "Accepted": [610, 585, 556, 544, 530, 515, 497, 451],
    "Uncertain": [18, 43, 72, 84, 98, 113, 131, 177],
    "Accepted Correct": [578, 558, 541, 533, 520, 506, 491, 450],
    "Accepted Incorrect": [32, 27, 15, 11, 10, 9, 6, 1],
    "Coverage (%)": [97.13, 93.15, 88.54, 86.62, 84.39, 82.01, 79.14, 71.82],
    "Accepted Accuracy (%)": [94.75, 95.38, 97.30, 97.98, 98.11, 98.25, 98.79, 99.78]
})

print("Custom Baseline CONFIDENCE THRESHOLD ANALYSIS")
print("=" * 80)
print(confidence_threshold_results.to_string(index=False))

print("\nSELECTED POLICY")
print("=" * 80)
print("Confidence threshold: 70%")
print("Accepted predictions: 556")
print("Uncertain predictions: 72")
print("Accepted correct: 541")
print("Accepted incorrect: 15")
print("Coverage: 88.54%")
print("Accepted accuracy: 97.30%")
````

````python
# PURPOSE: Visualize the coverage-versus-accuracy trade-off across confidence thresholds and highlight the selected 70% operating point.

import matplotlib.pyplot as plt

thresholds = confidence_threshold_results["Threshold"]
coverage = confidence_threshold_results["Coverage (%)"]
accepted_accuracy = confidence_threshold_results["Accepted Accuracy (%)"]
threshold_labels = list(thresholds)
x_positions = range(len(threshold_labels))
selected_idx = threshold_labels.index("70%")

plt.figure(figsize=(10, 6))

plt.plot(x_positions, coverage, marker="o", label="Coverage (%)", color="#2f6f8f")
plt.plot(x_positions, accepted_accuracy, marker="o", label="Accepted Accuracy (%)", color="#2f5d3a")

plt.axvline(x=selected_idx, color="gray", linestyle="--", linewidth=1.5, label="Selected threshold (70%)")

plt.annotate(
    f"{coverage[selected_idx]:.2f}%",
    xy=(selected_idx, coverage[selected_idx]),
    xytext=(selected_idx + 0.15, coverage[selected_idx] + 2),
    fontsize=9,
    color="#2f6f8f"
)
plt.annotate(
    f"{accepted_accuracy[selected_idx]:.2f}%",
    xy=(selected_idx, accepted_accuracy[selected_idx]),
    xytext=(selected_idx + 0.15, accepted_accuracy[selected_idx] - 4),
    fontsize=9,
    color="#2f5d3a"
)

plt.xticks(list(x_positions), threshold_labels)
plt.title("Confidence Threshold: Coverage vs. Accepted Accuracy")
plt.xlabel("Confidence Threshold")
plt.ylabel("Percentage (%)")
plt.ylim(60, 105)
plt.legend(loc="lower left")
plt.tight_layout()
plt.show()
````

## Confidence Threshold Findings

The confidence analysis demonstrates a trade-off between prediction coverage and accepted-prediction accuracy.

At the selected **70% confidence threshold**, the system accepts 556 of the 628 test predictions, corresponding to **88.54% coverage**. The remaining **72 predictions (11.46%) are treated as uncertain** rather than being presented with the same level of confidence. Of the 556 accepted predictions, 541 are correct and 15 are incorrect, giving an accepted accuracy of **97.30%**.

Increasing the confidence threshold further improves the accuracy of accepted predictions but reduces coverage. For example, a 95% threshold produces **99.78% accepted accuracy**, but only **71.82% of the test predictions are accepted**.

The 70% threshold was therefore retained as a practical balance between accepting useful predictions and identifying predictions that should be treated as uncertain.

Importantly, confidence should not be interpreted as a probability of guaranteed correctness. The uncertainty mechanism is a risk-reduction measure that helps prevent low-confidence predictions from being presented as reliable diagnoses.

The remaining accepted errors also demonstrate why the system requires appropriate caution in communicating predictions, particularly for visually similar disease classes such as Blight and Gray_Leaf_Spot.
